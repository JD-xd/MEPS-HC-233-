```python
# Function for % users
def percent_users(series):
    return (series > 0).mean() * 100

# Service Metrics
services = {
    "Office-Based (OB)": "OBTOTV21",
    "Outpatient (OP)": "OPTOTV21",
    "Emergency Room (ER)": "ERTOT21",
    "Inpatient Hospitalization (IP)": "IPDIS21",
    "Dental Visits (DV)": "DVTOT21",
    "Home Health (HH)": "HHTOTD21",
    "Prescriptions (RX)": "RXTOT21"
}

results = []

for service, col in services.items():
    if col in df.columns:
        total = df[col].sum()
        avg = df[col].mean()
        pct_users = percent_users(df[col])
        results.append([service, col, total, avg, pct_users])
    else:
        results.append([service, col, "Not Found", "Not Found", "Not Found"])

# Convert to DataFrame
summary_df = pd.DataFrame(results, columns=["Service", "Column Used", "Total Use", "Average per Person", "% Users"])

print(summary_df)

# 🔹 ER Usage by Age Group
df['AGE_GROUP'] = pd.cut(df['AGE21X'], bins=[0, 17, 44, 64, 120], labels=['0-17', '18-44', '45-64', '65+'])
er_by_age = df.groupby('AGE_GROUP')['ERTOT21'].apply(percent_users)

print("\n--- ER Usage by Age Group ---")
print(er_by_age)

# 🔹 Inpatient: Average Nights
if "IPNGTD21" in df.columns:
    print(f"\nAverage Nights in Hospital (among all people): {df['IPNGTD21'].mean():.2f}")
```

```python
# Helper function
def percent_users(series):
    return (series > 0).mean() * 100

# ================================
# AGE-WISE UTILIZATION
# ================================
df['AGE_GROUP'] = pd.cut(df['AGE21X'], 
                         bins=[0, 17, 44, 64, 120],
                         labels=['Children (0–17)', 'Adults (18–44)', 'Middle Age (45–64)', 'Elderly (65+)'])

age_group_util = df.groupby('AGE_GROUP').agg({
    'OBTOTV21': 'mean',     # Doctor visits
    'ERTOT21': percent_users, # % with ER visit
    'IPDIS21': percent_users, # % hospitalized
    'RXTOT21': percent_users  # % using prescriptions
}).rename(columns={
    'OBTOTV21': 'Avg Doctor Visits',
    'ERTOT21': '% ER Users',
    'IPDIS21': '% Hospitalized',
    'RXTOT21': '% Using Prescriptions'
})

print("\n=== Age-wise Utilization ===")
print(age_group_util)

# ================================
# GENDER-WISE UTILIZATION
# ================================
gender_map = {1: 'Male', 2: 'Female'}
df['GENDER'] = df['SEX'].map(gender_map)

gender_util = df.groupby('GENDER').agg({
    'OBTOTV21': 'mean',
    'ERTOT21': percent_users,
    'RXTOT21': percent_users
}).rename(columns={
    'OBTOTV21': 'Avg Doctor Visits',
    'ERTOT21': '% ER Users',
    'RXTOT21': '% Using Prescriptions'
})

print("\n=== Gender-wise Utilization ===")
print(gender_util)

# ================================
# POVERTY CATEGORY UTILIZATION
# ================================
pov_map = {
    1: 'Poor',
    2: 'Near Poor',
    3: 'Low Income',
    4: 'Middle Income',
    5: 'High Income'
}
df['POV_GROUP'] = df['POVCAT21'].map(pov_map)

pov_util = df.groupby('POV_GROUP').agg({
    'OBTOTV21': 'mean',
    'ERTOT21': percent_users,
    'IPDIS21': percent_users
}).rename(columns={
    'OBTOTV21': 'Avg Doctor Visits',
    'ERTOT21': '% ER Users',
    'IPDIS21': '% Hospitalized'
})

print("\n=== Income/Poverty-wise Utilization ===")
print(pov_util)

# ================================
# INSURANCE COVERAGE EFFECT
# ================================
# We'll use INSCOPE (1=insured, 2=not insured)
ins_map = {1: 'Insured', 2: 'Uninsured'}
df['INS_STATUS'] = df['INSCOPE'].map(ins_map)

insurance_util = df.groupby('INS_STATUS').agg({
    'OBTOTV21': 'mean',
    'ERTOT21': percent_users
}).rename(columns={
    'OBTOTV21': 'Avg Doctor Visits',
    'ERTOT21': '% ER Users'
})

print("\n=== Insurance Coverage Impact ===")
print(insurance_util)
```

```python
import pandas as pd
import numpy as np

# Load dataset
df = pd.read_csv("h233.csv")

# =============================
# Step 1: Define high utilizers
# =============================

# We'll use total expenditure as measure of utilization
df['TOTEXP21'] = df['TOTEXP21'].fillna(0)

# 95th percentile cutoff
cutoff = np.percentile(df['TOTEXP21'], 95)

# Label groups
df['UTIL_GROUP'] = np.where(df['TOTEXP21'] >= cutoff, 'High Utilizer (Top 5%)', 'Others (Bottom 95%)')

print(f"Cutoff for Top 5% = ${cutoff:,.2f}")

# =============================
# Step 2: Compare demographics
# =============================

# Age
df['AGE_GROUP'] = pd.cut(df['AGE21X'], 
                         bins=[0, 17, 44, 64, 120],
                         labels=['Children (0–17)', 'Adults (18–44)', 'Middle Age (45–64)', 'Elderly (65+)'])

# Gender
gender_map = {1: 'Male', 2: 'Female'}
df['GENDER'] = df['SEX'].map(gender_map)

# Poverty group
pov_map = {
    1: 'Poor',
    2: 'Near Poor',
    3: 'Low Income',
    4: 'Middle Income',
    5: 'High Income'
}
df['POV_GROUP'] = df['POVCAT21'].map(pov_map)

# Insurance
ins_map = {1: 'Insured', 2: 'Uninsured'}
df['INS_STATUS'] = df['INSCOPE'].map(ins_map)

# =============================
# Step 3: Summary comparison
# =============================

summary = df.groupby('UTIL_GROUP').agg({
    'AGE21X': 'mean',
    'TOTEXP21': 'mean',
    'OBTOTV21': 'mean',
    'ERTOT21': 'mean',
    'IPDIS21': 'mean'
}).rename(columns={
    'AGE21X': 'Avg Age',
    'TOTEXP21': 'Avg Total Expenditure',
    'OBTOTV21': 'Avg Doctor Visits',
    'ERTOT21': 'Avg ER Visits',
    'IPDIS21': 'Avg Hospital Stays'
})

print("\n=== Utilization Summary: Top 5% vs Others ===")
print(summary)

# =============================
# Step 4: Distribution by demographics
# =============================

def get_distribution(col):
    return (df.groupby(['UTIL_GROUP', col])
              .size()
              .unstack(fill_value=0)
              .apply(lambda x: 100 * x / x.sum(), axis=1)
              .round(1))

print("\n=== Age Group Distribution ===")
print(get_distribution('AGE_GROUP'))

print("\n=== Gender Distribution ===")
print(get_distribution('GENDER'))

print("\n=== Poverty Category Distribution ===")
print(get_distribution('POV_GROUP'))

print("\n=== Insurance Status Distribution ===")
print(get_distribution('INS_STATUS'))

# =============================
# Step 5 (optional): Chronic Illness Indicator
# =============================

# Example columns: DIABDX (Diabetes), HYPEV (Hypertension), ASTHDX (Asthma)
chronic_cols = ['DIABDX', 'HYPEV', 'ASTHDX']
for col in chronic_cols:
    if col in df.columns:
        chronic_rate = df.groupby('UTIL_GROUP')[col].mean() * 100
        print(f"\n% with {col} by Utilization Group:")
        print(chronic_rate.round(1))
```

```python
import pandas as pd
import numpy as np

# Load dataset
df = pd.read_csv("h233.csv")

# Fill missing values
df = df.fillna(0)

# =============================
# Step 1: Define Utilization Metrics
# =============================

# Office-based doctor visits (OB)
if 'OBTOTV21' in df.columns:
    df['OB_VISITS'] = df['OBTOTV21']
else:
    raise KeyError("Column 'OBTOTV21' not found. Please verify dataset.")

# Emergency Room visits (ER)
if 'ERTOT21' in df.columns:
    df['ER_VISITS'] = df['ERTOT21']
else:
    raise KeyError("Column 'ERTOT21' not found. Please verify dataset.")

# =============================
# Step 2: Classify Users
# =============================

def classify_usage(row):
    if row['ER_VISITS'] > 0 and row['OB_VISITS'] == 0:
        return 'ER-only User'
    elif row['OB_VISITS'] > 0 and row['ER_VISITS'] == 0:
        return 'Preventive-only User'
    elif row['OB_VISITS'] > 0 and row['ER_VISITS'] > 0:
        return 'Both ER + Preventive'
    else:
        return 'No Utilization'

df['USAGE_TYPE'] = df.apply(classify_usage, axis=1)

# =============================
# Step 3: Overall Percentages
# =============================

overall = df['USAGE_TYPE'].value_counts(normalize=True) * 100
print("=== Overall Preventive vs Emergency Reliance (%) ===")
print(overall.round(2))

# =============================
# Step 4: Add Demographic Columns
# =============================

# Age group
df['AGE_GROUP'] = pd.cut(df['AGE21X'],
                         bins=[0, 17, 44, 64, 120],
                         labels=['Children (0–17)', 'Adults (18–44)', 'Middle Age (45–64)', 'Elderly (65+)'])

# Gender
gender_map = {1: 'Male', 2: 'Female'}
df['GENDER'] = df['SEX'].map(gender_map)

# Poverty level
pov_map = {
    1: 'Poor',
    2: 'Near Poor',
    3: 'Low Income',
    4: 'Middle Income',
    5: 'High Income'
}
df['POV_GROUP'] = df['POVCAT21'].map(pov_map)

# Insurance coverage
ins_map = {1: 'Insured', 2: 'Uninsured'}
df['INS_STATUS'] = df['INSCOPE'].map(ins_map)

# =============================
# Step 5: Group-wise Comparison
# =============================

def usage_distribution(col):
    return (
        df.groupby([col, 'USAGE_TYPE'])
          .size()
          .unstack(fill_value=0)
          .apply(lambda x: 100 * x / x.sum(), axis=1)
          .round(1)
    )

print("\n=== Usage Type by Age Group (%) ===")
print(usage_distribution('AGE_GROUP'))

print("\n=== Usage Type by Gender (%) ===")
print(usage_distribution('GENDER'))

print("\n=== Usage Type by Poverty Category (%) ===")
print(usage_distribution('POV_GROUP'))

print("\n=== Usage Type by Insurance Status (%) ===")
print(usage_distribution('INS_STATUS'))
```

```python
import pandas as pd
import numpy as np

# Load the dataset
df = pd.read_csv("h233.csv")
df = df.fillna(0)

# =============================
# Step 1: Define Utilization Components
# =============================
df['TOTAL_VISITS'] = (
    df.get('OBTOTV21', 0) +   # Office-based
    df.get('ERTOT21', 0) +    # ER visits
    df.get('IPDIS21', 0) +    # Inpatient stays
    df.get('RXTOT21', 0)      # Prescriptions
)

# =============================
# Step 2: Define Need Indicators
# =============================

# Chronic illness (binary flags)
chronic_vars = ['CANCERDX', 'ASTHDX', 'DIABDX', 'CHDDX', 'HIBPDX']  # adjust based on dataset availability
for col in chronic_vars:
    if col not in df.columns:
        df[col] = 0

df['CHRONIC_COUNT'] = df[chronic_vars].sum(axis=1)
df['HAS_CHRONIC'] = np.where(df['CHRONIC_COUNT'] > 0, 1, 0)

# Poverty category (1 = Poor → 5 = High Income)
df['POV_GROUP'] = df['POVCAT21'].replace({
    1: 'Poor',
    2: 'Near Poor',
    3: 'Low Income',
    4: 'Middle Income',
    5: 'High Income'
})

# Insurance
df['INS_STATUS'] = df['INSCOPE'].replace({1: 'Insured', 2: 'Uninsured'})

# Age group
df['AGE_GROUP'] = pd.cut(df['AGE21X'],
                         bins=[0, 17, 44, 64, 120],
                         labels=['Children', 'Adults', 'Middle Age', 'Elderly'])

# Gender
df['GENDER'] = df['SEX'].replace({1: 'Male', 2: 'Female'})

# =============================
# Step 3: Compute Utilization Percentiles
# =============================
low_threshold = df['TOTAL_VISITS'].quantile(0.20)
high_threshold = df['TOTAL_VISITS'].quantile(0.90)

def utilization_group(row):
    if row['TOTAL_VISITS'] >= high_threshold:
        return 'High Utilizer'
    elif row['TOTAL_VISITS'] <= low_threshold and row['HAS_CHRONIC'] == 1:
        return 'Under-utilizer'
    else:
        return 'Normal'

df['UTIL_GROUP'] = df.apply(utilization_group, axis=1)

# =============================
# Step 4: Profile Each Group
# =============================
profile = (
    df.groupby('UTIL_GROUP')
      .agg({
          'AGE21X': 'mean',
          'CHRONIC_COUNT': 'mean',
          'TOTAL_VISITS': 'mean',
          'INSCOPE': lambda x: (x == 2).mean() * 100,
          'POVCAT21': lambda x: (x <= 2).mean() * 100,
      })
      .rename(columns={
          'AGE21X': 'Avg Age',
          'CHRONIC_COUNT': 'Avg Chronic Conditions',
          'TOTAL_VISITS': 'Avg Visits',
          'INSCOPE': '% Uninsured',
          'POVCAT21': '% Poor/Low Income'
      })
      .round(2)
)

print("=== Utilization Inequality Summary ===")
print(profile)

# =============================
# Step 5: Group-level Patterns
# =============================
demographic_patterns = (
    df.groupby(['UTIL_GROUP', 'AGE_GROUP', 'POV_GROUP', 'INS_STATUS', 'GENDER'])
      .size()
      .unstack(fill_value=0)
)

print("\n=== Demographic Breakdown of Utilization Groups ===")
print(demographic_patterns)
```
